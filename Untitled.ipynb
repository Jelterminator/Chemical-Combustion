```julia
using Plots

x=-10:1e-1:10

function clipping(x)
    return log(1+exp(7x))/7
end

function tanhzero(x)
    return tanh(50x)^2
end

plot1= plot(x,x,label="Actual value")
plot!(plot1, x, log1p.(exp.(7*x))/7,label="Softplus clipped value")
plot!(plot1, x, tanhzero.(x), label="Tanh sign value")
plot!(plot1, x,clipping.(x).*tanhzero.(x), label="Zero-is-zero clipped value")
display(plot1)
savefig("softclip")

x=-0.1:1e-3:0.1

plot2=plot(x,x,label="Actual value")
plot!(plot2, x, clipping.(x),label="Softplus clipped value")
plot!(plot2, x,clipping.(x).*tanhzero.(x), label="Zero-is-zero clipped value")
display(plot2)
savefig("zeroclip")
```
