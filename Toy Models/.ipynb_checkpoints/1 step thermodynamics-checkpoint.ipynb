```julia
using DifferentialEquations
using Plots
include("Chemkin.jl")

# Initial species concentrations 0. T 1. N2, 2. O2, 3. CH4, 4. H2O, 5. CO2

species_names = ["N2", "O2", "CH4", "H2O", "CO2"]

X0 = zeros(length(species_names)+1)
X0[1] = 600 # K
X0[2:end] = 20.06*[7.57/10.57, 2/10.57, 1/10.57, 0, 0] #mol/m^3

# Stoichiometry: 2 O2 + CH4 -> 2 H2O + CO2
S = [0, -2, -1, 2, 1]  # N2 does not participate in the reaction

# Molar masses
M = [28.0134, 31.999, 16.04246, 18.01528, 44.01]*1e-3  # kg/mol

species_dict = load_chemical_data("CHEMKIN-THERMDAT.txt")
```

```julia
function Arrhenius(Y)
    T = Y[1]
    X = Y[2:end]
    # Constants from Westbrook & Dryer 1981
    n_CH4 = -0.3
    n_O2 = 1.3
    A = 1.3e8 * 1e6
    Ea = 48.4 * 4184 # kcal/mol to J/mol
    R = 8.314 # J/mol/K
    
    # Ensure positive concentrations for exponentiation
    CH4_concentration = max(X[3], 0)
    O2_concentration = max(X[2], 0)
    
    k = A * exp(-Ea / (R * T)) # rate constant
    r = k * CH4_concentration^n_CH4 * O2_concentration^n_O2 # reaction rate
    return r
end
```

```julia
# Temperature rate of change due to reaction enthalpy
function dT(X, r)
    h0_vec = []
    cp_vec = []
    for i in 1:length(X)-1
        append!(h0_vec, h0(X[1], species_dict[species_names[i]]))
        append!(cp_vec, species_cp(X[1], species_dict[species_names[i]]))
    end
    dH = sum(-r.*S.*h0_vec) # 1/s * J/mol
    c_p = sum(X[2:end].*cp_vec)/sum(X[2:end]) #J/molK
    dT = dH/c_p #J/mols * molK/J
    return dT[1]
end
```

```julia
# Derivative function for the ODE system
function f!(dX, X, p, t)
    r = Arrhenius(X)
    dX[1] = dT(X,r)
    dX[2:end] = S' * r  # Species concentrations change
end

# define
tend = 2.5e-6
tspan = (0, tend)

# define problem 
problem = ODEProblem(f!, X0, tspan)

# solve problem 
sol = solve(problem, alg_hints=[:stiff])
```

```julia
# Extract the solution arrays
t = sol.t
T = sol[1, :]
C = sol[2:end, :]'

# Create a plot for the temperature
plot1 = plot(t, T, xlabel="Time (s)", ylabel="Temperature (K)", title="Temperature Evolution")

display(plot1)

# Create a plot for the species concentrations
plot2 = plot(t, C, xlabel="Time (s)", ylabel="Concentration", title="Species Concentrations", label=["N2" "O2" "CH4" "H2O" "CO2"])

display(plot2)
```
